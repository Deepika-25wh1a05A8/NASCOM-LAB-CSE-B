Task 7: Calculate ∫₀⁵ L(x) dx.
L(x) = 5x − 0.2x²
24. Solve the integral manually.
25. Implement a numerical approximation using Python.
26. Explain what the area under the curve represents in this case.

```python
import sympy as sp

x = sp.symbols('x')

L = 5*x - 0.2*x**2

integral = sp.integrate(L, (x, 0, 5))

print("Integral =", integral)
print("Value =", float(integral))
```

```python
import numpy as np

def L(x):
    return 5*x - 0.2*x**2

x = np.linspace(0, 5, 1000)
y = L(x)

area = np.trapezoid(y, x)

print("Numerical approximation =", area)
```

```python
import numpy as np
import matplotlib.pyplot as plt

def L(x):
    return 5*x - 0.2*x**2

x = np.linspace(0, 5, 100)
y = L(x)

plt.plot(x, y, label="L(x) = 5x - 0.2x²")
plt.fill_between(x, y, alpha=0.3)

plt.xlabel("x")
plt.ylabel("Learning Benefit")
plt.title("Area Under Learning Benefit Curve")
plt.legend()
plt.show()
```

11. Part H – Correlation Analysis
Calculate the correlation between:
 Study Hours and Final Score
 Attendance and Final Score
 Assignment Score and Final Score
 Previous Score and Final Score

```python
print(df.corr(numeric_only=True))
```

```python
X = df[["study_hours", "attendance", "assignment_score"]].values
Y = df["final_score"].values
X = np.column_stack((np.ones(len(X)), X))
beta = np.linalg.inv(X.T @ X) @ X.T @ Y
print("Coefficients:")
print(beta)
```

```python
print("Average final score:", df["final_score"].mean())
print("Variation in final scores:", df["final_score"].std())
student = df.loc[df["final_score"].idxmax()]

print("Student with highest final score:")
print(student)
import seaborn as sns
import matplotlib.pyplot as plt

sns.scatterplot(data=df, x="study_hours", y="final_score")
plt.show()
sns.scatterplot(data=df, x="attendance", y="final_score")
plt.show()
corr = df.corr(numeric_only=True)["final_score"].drop("final_score")

print("Strongest correlation:")
print(corr.abs().idxmax())

print("Correlation value:")
print(corr[corr.abs().idxmax()])
```
