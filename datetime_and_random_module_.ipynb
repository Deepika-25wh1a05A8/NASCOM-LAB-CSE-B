```python
from datetime import datetime, date, timedelta
now = datetime.now()
print("Now:", now)
today = date.today()
print("Today's Date:", today)
dob = date(2007, 2, 6)
print("DOB:", dob)
yesterday = today - timedelta(days=1)
tomorrow = today + timedelta(days=1)
print("Yesterday:", yesterday)
print("Tomorrow:", tomorrow)
formatted = now.strftime("%d-%m-%Y %H:%M:%S")
print("Formatted DateTime:", formatted)
```

```python
import random
print("Random Int:", random.randint(2,34))
print("Random Float:", random.random())
colors = ['pink', 'green', 'yellow']
print("Random Color:", random.choice(colors))
random.shuffle(colors)
print("Shuffled List:", colors)
print("Sample:", random.sample(range(1, 30), 3))
```

```python
from collections import Counter
fruits=['apple','grapes','banana','kiwi']
count= Counter(fruits)
print("fruits count:",count)
```

```python
from collections import defaultdict
marks = defaultdict(lambda: 0)
marks['Math'] = 90
marks['Science'] = 80
print("English Marks (not set):", marks['English'])  # Returns 0
```

```python
from collections import namedtuple
Point = namedtuple('Point', 'x y')
p1 = Point(10, 20)
print("Point:", p1)
print("x:", p1.x, "y:", p1.y)
```

```python
from collections import deque
dq = deque([4,2,5,3,6])
dq.append(7)
dq.appendleft(0)
dq.pop()
dq.popleft()
print("Deque after operations:", dq)
```
